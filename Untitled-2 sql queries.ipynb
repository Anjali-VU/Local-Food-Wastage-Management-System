```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

conn = sqlite3.connect("food_wastage.db")
cursor = conn.cursor()
```

```python
q = "SELECT COUNT(*) AS total_providers FROM providers;"
df = pd.read_sql_query(q, conn)
df
```

```python
plt.bar(["Providers"], df["total_providers"])
plt.title("Total Providers")
plt.show()
```

```python
q = "SELECT COUNT(*) AS total_receivers FROM receivers;"
df = pd.read_sql_query(q, conn)
df
```

```python
plt.bar(["Receivers"], df["total_receivers"])
plt.title("Total Receivers")
plt.show()
```

```python
q = "SELECT COUNT(*) AS total_food FROM food_listings;"
df = pd.read_sql_query(q, conn)
df
```

```python
plt.bar(["Food Listings"], df["total_food"])
plt.title("Total Food Listings")
plt.show()
```

```python
q = "SELECT City, COUNT(*) AS count FROM providers GROUP BY City;"
df = pd.read_sql_query(q, conn)
df
```

```python
plt.bar(df["City"], df["count"])
plt.xticks(rotation=45)
plt.title("Providers by City")
plt.show()
```

```python
q = """
SELECT * 
FROM food_listings 
WHERE date(Expiry_Date) >= date('now');
"""

df = pd.read_sql_query(q, conn)
df
```

```python
q1 = "SELECT COUNT(*) AS Active FROM food_listings WHERE date(Expiry_Date) >= date('now');"
q2 = "SELECT COUNT(*) AS Expired FROM food_listings WHERE date(Expiry_Date) < date('now');"

active = pd.read_sql_query(q1, conn).iloc[0,0]
expired = pd.read_sql_query(q2, conn).iloc[0,0]

import matplotlib.pyplot as plt

plt.bar(["Active", "Expired"], [active, expired])
plt.title("Food Expiry Status")
plt.show()
```

```python
q = """
SELECT Food_Type, COUNT(*) AS count
FROM food_listings
GROUP BY Food_Type;
"""

df = pd.read_sql_query(q, conn)
df
```

```python
plt.bar(df["Food_Type"], df["count"])
plt.xticks(rotation=45)
plt.title("Food Type Distribution")
plt.show()
```

```python
q = """
SELECT Provider_ID, COUNT(*) AS count
FROM food_listings
GROUP BY Provider_ID
ORDER BY count DESC;
"""

df = pd.read_sql_query(q, conn)
df
```

```python
plt.bar(df["Provider_ID"].astype(str), df["count"])
plt.xticks(rotation=45)
plt.title("Provider Contribution")
plt.show()
```

```python
q = "SELECT Status, COUNT(*) AS count FROM claims GROUP BY Status;"
df = pd.read_sql_query(q, conn)
df
```

```python
plt.pie(df["count"], labels=df["Status"], autopct="%1.1f%%")
plt.title("Claims Status Distribution")
plt.show()
```

```python
q = """
SELECT Food_ID, COUNT(*) AS count
FROM claims
GROUP BY Food_ID
ORDER BY count DESC
LIMIT 5;
"""
df = pd.read_sql_query(q, conn)
df
```

```python
plt.bar(df["Food_ID"].astype(str), df["count"])
plt.title("Most Claimed Food")
plt.show()
```

```python
q = """
SELECT f.Provider_ID, COUNT(*) AS completed_claims
FROM claims c
JOIN food_listings f
ON c.Food_ID = f.Food_ID
WHERE c.Status = 'Completed'
GROUP BY f.Provider_ID
ORDER BY completed_claims DESC;
"""

df = pd.read_sql_query(q, conn)
df
```

```python
import matplotlib.pyplot as plt

plt.bar(df["Provider_ID"].astype(str), df["completed_claims"])
plt.xticks(rotation=45)
plt.title("Top Successful Providers")
plt.show()
```
